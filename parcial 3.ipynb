```sql
%sql
CREATE OR REPLACE VIEW workspace.electiva.orders_agosto_2011 AS
WITH ordenes_agosto AS (
    SELECT 
    p.ProductName AS productname,
    p.ModelName AS modelname,
    pu.user_name AS user_name,
    o.OrderDate AS orderdate,
    o.SalesOrderID AS salesorderid,
    pc.CategoryName AS categoryname,
    o.LineTotal AS linetotal,
    CASE 
            WHEN pc.CategoryName ILIKE 'Bikes' THEN 0.10
            WHEN pc.CategoryName ILIKE 'Clothing' THEN 0.15
            WHEN pc.CategoryName ILIKE 'Components' THEN 0.08
            WHEN pc.CategoryName ILIKE 'Accessories' THEN 0.20
            ELSE 0.00
        END AS `descuento_mitad_año`
        FROM read_files(
            '/Volumes/workspace/electiva/mi_volumen_datos/orders\\[1\\].csv',
            format => 'csv',
            header => true
        ) o
        INNER JOIN read_files(
            '/Volumes/workspace/electiva/mi_volumen_datos/products\\[1\\].csv',
            format => 'csv',
            header => true
        ) p ON o.ProductID = p.ProductID
        INNER JOIN read_files(
            '/Volumes/workspace/electiva/mi_volumen_datos/productsubcategories\\[1\\].csv',
            format => 'csv',
            header => true
        ) ps ON p.SubCategoryID = ps.SubCategoryID
        INNER JOIN read_files(
            '/Volumes/workspace/electiva/mi_volumen_datos/productcategories\\[1\\].csv',
            format => 'csv',
            header => true
        ) pc ON ps.CategoryID = pc.CategoryID
        INNER JOIN workspace.electiva.page_user pu ON o.CustomerID = pu.customerid
        WHERE o.OrderDate >= '2011-08-01' AND o.OrderDate <= '2011-08-31'
)
SELECT 
    productname,
    modelname,
    user_name,
    orderdate,
    salesorderid,
    categoryname,
    linetotal,
    `descuento_mitad_año`,
    linetotal - (linetotal * `descuento_mitad_año`)
    AS total_descuentos_aplicados
    FROM ordenes_agosto;
```

```sql
%sql
SELECT * FROM workspace.electiva.orders_agosto_2011
LIMIT 20;
```

```sql
%sql
CREATE OR REPLACE MATERIALIZED VIEW workspace.electiva.customer_class AS

WITH conteo_ordenes AS (
    SELECT
        CustomerID,
        COUNT(DISTINCT SalesOrderID) AS cantidad_ordenes
    FROM read_files(
        '/Volumes/workspace/electiva/mi_volumen_datos/orders\\[1\\].csv',
        format => 'csv',
        header => true
    )
    WHERE OrderDate >= DATE '2011-01-01'
      AND OrderDate < DATE '2012-01-01'
    GROUP BY CustomerID
)

SELECT
    pu.user_name,
    CASE
        WHEN COALESCE(co.cantidad_ordenes, 0) = 3 THEN 'Oro'
        WHEN COALESCE(co.cantidad_ordenes, 0) = 2 THEN 'Plata'
        ELSE 'Ocasional'
    END AS user_level,
    c.CustomerID AS customerid,
    c.FullName AS fullname,
    COALESCE(co.cantidad_ordenes, 0) AS cantidad_ordenes

FROM read_files(
    '/Volumes/workspace/electiva/mi_volumen_datos/customers\\[1\\].csv',
    format => 'csv',
    header => true,
    encoding => 'ISO-8859-1'
) c

INNER JOIN workspace.electiva.page_user pu
    ON c.CustomerID = pu.customerid

LEFT JOIN conteo_ordenes co
    ON c.CustomerID = co.CustomerID;
```

```sql
%sql
SELECT *
FROM workspace.electiva.customer_class
ORDER BY cantidad_ordenes DESC
LIMIT 10;
```

```sql
%sql
SELECT
    user_level,
    COUNT(*) AS total_clientes
FROM workspace.electiva.customer_class
GROUP BY user_level
ORDER BY user_level;
```
